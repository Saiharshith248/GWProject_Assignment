```python
import math

class MissionaryCannibalProblem(Problem):
    def __init__(self, M, C, Bnum):
        self.M = M
        self.C = C
        self.Bnum = Bnum

        self.expandedNodesCount = 0

        initial = tuple([self.M, self.C, 1])
        goal = tuple([0, 0, 0])

        super().__init__(initial, goal)
    
    def isValidState(self, state):
        state = list(state)

        m = state[0]
        c = state[1]

        return (m >= c and m >= 0 and c >= 0) or (m < c and m == 0)

    def actions(self, state):
        state = list(state)
        b = state[2]

        wrongM = state[0]
        wrongC = state[1]
        correctM = self.M - wrongM
        correctC = self.C - wrongC

        validActions = []
        for bnum in range(1, self.Bnum + 1):
            for m in range(0, bnum + 1):
                c = bnum - m

                boat = tuple([m, c])
                wrong = tuple([wrongM - m, wrongC - c]) if b == 1 else tuple([wrongM + m, wrongC + c])
                correct = tuple([correctM + m, correctC + c]) if b == 1 else tuple([correctM - m, correctC - c])

                if self.isValidState(boat) and self.isValidState(wrong) and self.isValidState(correct):
                    action = 'M' * m + 'C' * c
                    validActions.append(action)
        
        return validActions if len(validActions) > 0 else None
    
    def result(self, state, action):
        state = list(state)

        mBoat = action.count('M')
        cBoat = action.count('C')

        newM = state[0] - mBoat if state[2] == 1 else state[0] + mBoat
        newC = state[1] - cBoat if state[2] == 1 else state[1] + cBoat
        newB = 0 if state[2] == 1 else 1

        return tuple([newM, newC, newB])
    
    def path_cost(self, c, state1, action, state2):
        return super().path_cost(c, state1, action, state2)
    
    def h(self, node):
        return math.ceil((node.state[0] + node.state[1]) / self.Bnum)
```

```python
prob = MissionaryCannibalProblem(3, 3, 2)
result = recursive_best_first_search(prob, h=prob.h)

print(result.path())
print(len(result.path()) - 1)
```

```python
prob = MissionaryCannibalProblem(7, 7, 4)
result = astar_search(prob, h=prob.h, display=True)

print(result.path())
print(len(result.path()) - 1)
```

```python
class RoadMap(GraphProblem):
    def __init__(self, initial, goal, graph, sld):
        super().__init__(initial, goal, graph)
        self.sld = sld

        self.expandedNodesCount = 0
        self.isConsistentHeuristic = True
    
    def path_cost(self, cost_so_far, A, action, B):
        self.isConsistentHeuristic = self.isConsistentHeuristic and (self.graph.get(A, B) + self.sld[B] >= self.sld[A])

        return super().path_cost(cost_so_far, A, action, B)
    
    def h(self, node):
        return self.sld[node.state]
```

```python
with open('cities_adjacency.txt', 'r') as file:
    lines = file.readlines()

cities_adj = dict()

for line in lines:
    words = line.strip().split('<->')
    city1 = words[0].strip()

    words = words[1].strip().split(':')
    city2 = words[0].strip()
    dist = int(words[1].strip())

    if city1 in cities_adj:
        cities_adj[city1][city2] = int(dist)
    else:
        cities_adj[city1] = { city2: int(dist) }

with open('cities_heuristic.txt', 'r') as file:
    lines = file.readlines()

cities_h = dict()

for line in lines:
    words = line.strip().split('->')
    city1 = words[0].strip()

    words = words[1].strip().split(':')
    city2 = words[0].strip()
    dist = int(words[1].strip())

    cities_h[city1] = int(dist)
```

```python
graph = UndirectedGraph(cities_adj)
sld = cities_h
```

```python
prob = RoadMap('Las Vegas', 'Dallas', graph, sld)
result = astar_search(prob, h=prob.h, display=True)

print(result.path())
print(len(result.path()) - 1)

isConsistent = "consistent" if prob.isConsistentHeuristic else "not consistent"
print("Heuristic is " + isConsistent)
```
